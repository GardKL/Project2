```python
   import jax
   import numpy as np
   import matplotlib.pyplot as plt
   from sklearn.model_selection import train_test_split
   
   from neural_network import create_layers_batch, feed_forward_batch
   from activations import sigmoid, sigmoid_der, linear, linear_der
   from train import train_gd, train_sgd
   from regression import make_data, polynomial_features, ols, fit_predict, MSE, runge 
```

Setting up the data in the same way as from project 1

```python
x, y = make_data(n=100, sigma = 0.1, seed = 2026)
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.2, random_state = 2026 )
y_train_pred, y_test_pred, theta = fit_predict(x_train, x_test, y_train, degree=10)
print("OLS degree 10: train MSE", MSE(y_train, y_train_pred), " test MSE", MSE(y_test, y_test_pred))
```

```python
from sklearn.preprocessing import StandardScaler

X_train = x_train[:, None]          # form (80, 1)
X_test  = x_test[:, None]           # form (20, 1)
Y_train = y_train[:, None]
Y_test  = y_test[:, None]

scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)   # lærer snitt og std fra treningsdata, og skalerer
X_test_s  = scaler.transform(X_test)        # bruker de SAMME tallene på testdata
```

```python
layers = create_layers_batch(1,[50,1], np.random.default_rng(2026))

layers, history = train_gd(X_train_s,Y_train,layers, [sigmoid,linear], [sigmoid_der, linear_der]
                            ,gamma = 0.2, n_steps=10000)

Y_test_pred = feed_forward_batch(X_test_s,layers,[sigmoid,linear])
print("train MSE", history[-1], " test MSE", MSE(Y_test, Y_test_pred))
```

```python
n = len(X_train_s)

layers_gd = create_layers_batch(1, [50,1], np.random.default_rng(2026))
layers_gd, history_gd = train_gd(X_train_s, Y_train,layers_gd, [sigmoid,linear], [sigmoid_der,linear_der], gamma = 0.2, n_steps = 1000)

layers_sgd = create_layers_batch(1, [50,1], np.random.default_rng(2026))
layers_sgd, hist_sgd = train_sgd(X_train_s, Y_train, layers_sgd,
                                 [sigmoid, linear], [sigmoid_der, linear_der],
                                 method="plain", gamma=0.2, n_epochs=1000, batch_size=n)

print(history_gd[-1], hist_sgd[-1])
```
